Create a function to save data into a CSV file

```python
def save_df_to_csv(df, filename, index=False, sep=',', encoding='utf-8'):
    """
    Save a pandas DataFrame to a CSV file.
    Ensures the target directory exists.
    Returns the path to the written file.
    """
    import os
    parent = os.path.dirname(filename)
    if parent:
        os.makedirs(parent, exist_ok=True)
    df.to_csv(filename, index=index, sep=sep, encoding=encoding)
    return filename 
```

```python
from vnstock import Listing, Quote
import os
```

```python
hpgQuote = Quote(symbol='HPG', source='vci')
hpgDf = hpgQuote.history(start="2020-01-01", end='2025-10-31', interval='1D')
csv_path = save_df_to_csv(hpgDf, 'data/hpg_history.csv')
print(f"Data saved to {csv_path}")
```

```python
hpgQuote = Quote(symbol='VN30', source='vci')
hpgDf = hpgQuote.history(start="2020-01-01", end='2025-10-31', interval='1D')
csv_path = save_df_to_csv(hpgDf, 'data/vn30_history.csv')
print(f"Data saved to {csv_path}")
```
